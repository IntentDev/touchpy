```python
import touchpy as tp
import concurrent.futures as cf
tp.init_logging(tp.LogLevel.DEBUG, console=False, file=True)
comp = tp.Comp(flags=tp.CompFlags.INTERNAL_TIME_ASYNC)

future = cf.Future()
def on_loaded(comp, info):
	future.set_result(True)
	# print('Loaded')


data = {}
comp.set_on_loaded_callback(on_loaded, data)

comp.load_tox('MyComponent.tox')

```
