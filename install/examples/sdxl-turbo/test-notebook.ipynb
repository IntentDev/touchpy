```python
# imports
import numpy as np
import torch
from pathlib import Path
import sys

localImportPath = Path.cwd().parents[2] / 'modules'
if str(localImportPath) not in sys.path:
	sys.path.insert(0,str(localImportPath))	

import touchpy as tp

torch.cuda.set_device(0)
```

```python
# set tox to load
comp = tp.Comp("tox/yolo.tox", flags=tp.CompFlags.INTERNAL_TIME_ASYNC)
```

```python
def on_layout_change(comp, user_data):
    pass

comp.set_on_layout_change_callback(on_layout_change, {})
```

```python
user_data = {'frame': 0}

# callback doesn't work with primitive global types, probably because they are passed by value
# rather than by reference in python
# frame = 0 

# this will work because it's a dictionary and the data is passed by reference
local_data = {'frame': 0}

def on_frame(comp, user_data):

	# print("frame callback")	# printing here is not a good idea it will print a lot in the notebook :)
	#read Out TOPs from the tox we loaded
	webcam = comp.out_tops[0].as_tensor()

	
	canny = comp.out_tops[1].as_tensor()
	# print(canny.shape)
	######### Process and Copy for next frame (Fast) ###################
	comp.start_next_frame()	

	comp.in_tops[0].from_tensor(webcam)
	
	user_data['frame'] += 1
	# frame += 1
	local_data['frame'] += 1
	

comp.set_on_frame_callback(on_frame, user_data)
```

```python
comp.start()
```
